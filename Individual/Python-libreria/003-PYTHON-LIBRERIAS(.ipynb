### Ejercicio 2 – Dataset simple con Pandas

```
import pandas as pd

nombre = [
    ["Ana", 8], 
    ["Juan",6],
    ["Carla",9],
    ["Pedro",5],
    ["Lucia",7]
]
df = pd.DataFrame(nombre, columns=["Nombre","Nota"])
print(df)

n_prom = df["Nota"].mean()
print("El promedio de las notas es: ", n_prom)

n_max = df["Nota"].max()
print("La nota máxima es: ", n_max)

n_min = df["Nota"].min()
print("La nota mínima es: ", n_min)
```
